# EDA (Análisis Exploratorio)

Objetivo: describir datos limpios, métricas clave y gráficos básicos.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from src.utils.reporting import write_report

df = pd.read_csv('data/processed/clean_data.csv')
df.head()
```

```python
df.describe(include='all').T.head(20)
```

```python
# DQ básica
nulls = df.isnull().mean().sort_values(ascending=False).head(20)
write_report('reports/eda_report.md','Reporte EDA', {'rows': len(df), 'null_top20': nulls.to_dict()})
nulls.head(10)
```

```python
# Gráfico simple si hay campos de monto/cantidad
num_cols = [c for c in df.columns if df[c].dtype!='O']
if num_cols:
    sns.histplot(df[num_cols[0]], kde=True)
    plt.show()
```
